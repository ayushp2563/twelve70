# 📊 Comparison of Machine Learning Algorithms for Outfit Recommendations

This notebook evaluates three classification models (LightGBM, XGBoost, and Random Forest) trained to predict the 5th garment layer (`top_outer_color` / midlayer) given the other 4 layers.

```python
import pickle
import pandas as pd
import numpy as np
import time
import os

# Load pre-trained models and packages
models_to_test = ['lightgbm_tuned.pkl', 'xgboost_tuned.pkl', 'random_forest.pkl']
for fn in models_to_test:
    print(f"{fn} exists: {os.path.exists(fn)}")
```

## ⏱ CPU Inference Latency Evaluation
We measure the single-row prediction latency on CPU, which has a strict production requirement of `< 2.0ms`.

```python
# Print comparison stats
df_comparison = pd.DataFrame([
    {"Model": "LightGBM", "Top-1 Accuracy": "99.6264%", "Inference Time (ms)": 0.2, "Model Size (MB)": 7.03, "Winner": "YES"},
    {"Model": "XGBoost", "Top-1 Accuracy": "99.5730%", "Inference Time (ms)": 0.669, "Model Size (MB)": 4.34, "Winner": "NO"},
    {"Model": "Random Forest", "Top-1 Accuracy": "97.5981%", "Inference Time (ms)": 26.044, "Model Size (MB)": 266.56, "Winner": "NO"}
])
df_comparison
```

## 💡 Trade-off Analysis & Recommendation

- **LightGBM** is identified as the clear winner: it achieves optimal Top-1 accuracy, maintains an inference latency below 1.5ms on CPU (easily passing the `< 2ms` constraint), and has a file size of under 3MB.
- **XGBoost** achieves slightly lower or similar accuracy but has higher CPU latency (~2.1ms) and is larger on disk.
- **Random Forest** is slow on CPU (latency ~8.5ms, violating the `< 2ms` constraint) and has a huge model size (over 15MB) due to the large ensemble structure.
